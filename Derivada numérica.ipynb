```python
# Função
def f(x):
    return x**2

# Ponto onde queremos a derivada
x = 2

# Passo
h = 0.001

# Diferença forward
df_forward = (f(x + h) - f(x)) / h

# Diferença backward
df_backward = (f(x) - f(x - h)) / h

# Diferença central
df_central = (f(x + h) - f(x - h)) / (2*h)

print("Forward:", df_forward)
print("Backward:", df_backward)
print("Central:", df_central)

print("Valor real:", 2*x)
```

```python
# Nome: EDO simples por diferença finita
# O que faz? Resolve dT/dx = -kT usando discretização
# Qual caminho aborda? Substitui derivada por diferença progressiva

# Dados do problema
k = 0.5           # constante
dx = 0.1          # passo em x
x_final = 1.0     # até onde queremos calcular
T0 = 100          # condição inicial T(0)

# Número de passos
n_passos = int(x_final / dx)

# Listas para guardar resultados
x_valores = [0]
T_valores = [T0]

# Loop de cálculo
for i in range(n_passos):
    T_i = T_valores[i]                         # valor atual de T
    T_prox = T_i * (1 - k*dx)                  # fórmula discretizada
    x_prox = x_valores[i] + dx                 # próximo ponto em x

    T_valores.append(T_prox)                   # guarda novo T
    x_valores.append(x_prox)                   # guarda novo x

# Mostrar resultados
for i in range(len(x_valores)):
    print(f"x = {x_valores[i]:.1f}, T = {T_valores[i]:.4f}")
```
