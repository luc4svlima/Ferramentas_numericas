```python
# Nome: Bisseção 
# O que faz? Encontrar o valor de x para zerar a função f(x) 
# Qual caminho aborda? É um método que encontra a raiz reduzindo um intervalo pela metade repetidamente

# Recomendação: 
# Problemas com intervalos físicos conhecidos como Temperatura, Pressão, Fração de algum estado (vapor, liquido) 
# Derivada difícil ou inexistente
# Função não suave / com ruído
# Preciso de estabilidade, não velocidade
# Quero garantir convergência
# Não confio no comportamento da função

# Como funciona? 
# 1 - Defina um intervalo [a,b] tal que f(a).f(b)<0
# 2 - Calcule o ponto médio xm = a + b/2 
# 3 - Avaliar a função em f(xm) 
# 4 - Se estiver entre a e xm -> novo B       Se estiver entre xm e b -> novo a 
# 5 - Repetir até b - a/2   < tolerancia 
```

```python
def bissecao(f, a, b, tolerancia=1e-6, max_iter=100):  # Define a função "bissecao" com entradas: função f, intervalo [a,b], tolerância e máximo de iterações
    
    if f(a) * f(b) > 0:  # Verifica se NÃO há troca de sinal no intervalo
        raise ValueError("O intervalo [a,b] não possui troca de sinal.")  # Se não houver troca de sinal, interrompe com erro

    for i in range(max_iter):  # Loop que repete o método até o número máximo de iterações
        xm = (a + b) / 2  # Calcula o ponto médio do intervalo (coração da bisseção)
        erro = abs(b - a) / 2  # Calcula o erro atual como metade do tamanho do intervalo

        if f(xm) == 0 or erro < tolerancia:  # Critério de parada: raiz exata OU erro pequeno o suficiente
            return {  # Retorna um dicionário com os resultados
                "raiz": xm,  # Valor aproximado da raiz
                "iteracoes": i + 1,  # Número de iterações realizadas (i começa em 0, por isso +1)
                "erro": erro  # Erro final
            }

        if f(a) * f(xm) < 0:  # Verifica se a raiz está entre a e xm (troca de sinal nesse intervalo)
            b = xm  # Atualiza o limite superior → novo intervalo é [a, xm]
        else:
            a = xm  # Caso contrário, a raiz está entre xm e b → novo intervalo é [xm, b]

    return {  # Caso atinja o número máximo de iterações sem convergir
        "raiz": xm,  # Retorna a última estimativa da raiz
        "iteracoes": max_iter,  # Retorna o número máximo de iterações
        "erro": erro  # Retorna o erro final
    }
```

```python
def f(x):  # Define a função do problema (exemplo)
    return x**2 - 4  # x**2 significa x ao quadrado → função f(x) = x² - 4


resultado = bissecao(f, a=1, b=3, tolerancia=1e-6, max_iter=100)  
# Chama o método da bisseção:
# f → função definida acima
# a=1, b=3 → intervalo onde sabemos que existe raiz
# tolerancia → erro aceitável
# max_iter → limite de repetições


print("Raiz aproximada:", resultado["raiz"])  
# print → mostra na tela
# resultado["raiz"] → acessa o valor da chave "raiz" dentro do dicionário

print("Número de iterações:", resultado["iteracoes"])  
# Mostra quantas vezes o método precisou repetir

print("Erro final:", resultado["erro"])  
# Mostra o erro da solução encontrada
```
